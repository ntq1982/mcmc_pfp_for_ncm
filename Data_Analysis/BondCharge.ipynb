```python
'''
    This program plots Bader charge of metals in different NCM compounds.

    Extract: Data/OxidationStates.tgz to access the oxidation state data.

    Created on Dec 12, 2024 at RISM (Shinshu University)
    Last update: Aug 03, 2026 16:00 JST

    Copyright © 2024-2026 Quang Nguyen. All rights reserved.
'''

import os
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import ticker
from matplotlib.patches import Patch

# Compounds and charge data
os.chdir("./Data/OxidationStates/")
compound_labels = ['NCM523@0K', 'NCM523@50K', 'NCM523@600K', 'NCM523@SQS',
                   'NCM622@0K', 'NCM622@50K', 'NCM622@600K', 'NCM622@SQS',
                   'NCM811@0K', 'NCM811@50K', 'NCM811@600K', 'NCM811@SQS']
charge_files    = ['NCM523_Stable_Charges.dat', 'NCM523_50K_Charges.dat', 'NCM523_600K_Charges.dat', 'NCM523_SQS_Charges.dat',
                   'NCM622_Stable_Charges.dat', 'NCM622_50K_Charges.dat', 'NCM622_600K_Charges.dat', 'NCM622_SQS_Charges.dat',
                   'NCM811_Stable_Charges.dat', 'NCM811_50K_Charges.dat', 'NCM811_600K_Charges.dat', 'NCM811_SQS_Charges.dat']

# Format x-ticks to have multi-line labels
formatted_labels = [label.replace('@', '\n@') for label in compound_labels]

# Read data from each file
charge_data = {}
averages = {'Ni': [], 'Co': [], 'Mn': []}
for compound, file in zip(compound_labels, charge_files):
    charge_data[compound] = {'Ni': [], 'Co': [], 'Mn': []}
    with open(file, 'r') as f:
        for line in f:
            parts = line.split()
            if parts[0] in ['Ni', 'Co', 'Mn']:
                element = parts[0]
                charges = list(map(float, parts[1:]))
                charge_data[compound][element].extend(charges)
    for element in ['Ni', 'Co', 'Mn']:
        averages[element].append(np.mean(charge_data[compound][element]))

# Violin plot for charges
colors  = {'Ni': '#ff7f0e', 'Co': '#d62728', 'Mn': '#1f77b4'}
offsets = {'Ni': 0.0, 'Co': 0.0, 'Mn': 0.0}
plt.figure(figsize=(11.2, 5.6))
plt.gca().yaxis.set_major_formatter(ticker.FormatStrFormatter('%3.1f'))
plt.gca().yaxis.set_major_locator(ticker.MultipleLocator(0.2))
plt.gca().yaxis.set_minor_locator(ticker.MultipleLocator(0.1))
for element in ['Ni', 'Co', 'Mn']:
    data_for_element = []
    positions = []
    for i, compound in enumerate(compound_labels):
        data_for_element.append(charge_data[compound][element])
        positions.append(i + 1 + offsets[element])
    vp = plt.violinplot(data_for_element, positions=positions, widths=0.4, showmeans=False, showmedians=False, showextrema=False)
    for body in vp['bodies']:
        body.set_facecolor(colors[element])
        body.set_edgecolor('black')
        body.set_alpha(0.75)
legend_elements = [Patch(facecolor=colors['Ni'], edgecolor='black', label='Ni'),
                   Patch(facecolor=colors['Co'], edgecolor='black', label='Co'),
                   Patch(facecolor=colors['Mn'], edgecolor='black', label='Mn'),]

# Plot and connect average values for each compound (0K, 50K, 600K, SQS)
compound_indices = {'NCM523': [0, 1, 2, 3], 'NCM622': [4, 5, 6, 7], 'NCM811': [8, 9, 10, 11]}
for element in ['Ni', 'Co', 'Mn']:
    x_vals = [i + 1 + offsets[element] for i in range(len(compound_labels))]
    plt.scatter(x_vals, averages[element], color='black', s=50, zorder=5)
for element in ['Ni', 'Co', 'Mn']:
    for compound, indices in compound_indices.items():
        x_vals = [i + 1 + offsets[element] for i in indices]
        y_vals = [averages[element][i] for i in indices]
        plt.plot(x_vals, y_vals, color='black', linewidth=1)

# Other plot decoration
plt.ylim([1.0 - 0.2/5, 2.2 + 0.2/5])
plt.xticks(range(1, len(compound_labels) + 1), formatted_labels, rotation=45, fontsize=18)
plt.yticks(fontsize=24)
plt.ylabel('Bader Charge', fontsize=26)
plt.legend(handles=legend_elements, loc='upper right', ncol=3, frameon=False, handlelength=0.8, fontsize=20)
plt.tight_layout()
plt.savefig('NCM_Charges_0K_50K_600K_SQS_violin.png')
plt.show()
```

```python
'''
    This program plot metal-oxygen bond lengths in different NCM compounds.
    
    Extract: Data/OxidationStates.tgz to access the oxidation state data.

    Created on Dec 12, 2024 at RISM (Shinshu University)
    Last update: Aug 03, 2026 16:00 JST

    Copyright © 2024-2026 Quang Nguyen. All rights reserved.
'''

import os
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import ticker
from matplotlib.patches import Patch

# Compounds and bond length data
os.chdir("./Data/OxidationStates/")
compound_labels = ['NCM523@0K', 'NCM523@50K', 'NCM523@600K', 'NCM523@SQS',
                   'NCM622@0K', 'NCM622@50K', 'NCM622@600K', 'NCM622@SQS',
                   'NCM811@0K', 'NCM811@50K', 'NCM811@600K', 'NCM811@SQS']
bond_files      = ['NCM523_Stable_Bonds.dat', 'NCM523_50K_Bonds.dat', 'NCM523_600K_Bonds.dat', 'NCM523_SQS_Bonds.dat',
                   'NCM622_Stable_Bonds.dat', 'NCM622_50K_Bonds.dat', 'NCM622_600K_Bonds.dat', 'NCM622_SQS_Bonds.dat',
                   'NCM811_Stable_Bonds.dat', 'NCM811_50K_Bonds.dat', 'NCM811_600K_Bonds.dat', 'NCM811_SQS_Bonds.dat']

# Format x-ticks to have multi-line labels
formatted_labels = [label.replace('@', '\n@') for label in compound_labels]

# Read data from each file
bond_data = {}
averages = {'Ni-O': [], 'Co-O': [], 'Mn-O': []}
for compound, file in zip(compound_labels, bond_files):
    bond_data[compound] = {'Ni-O': [], 'Co-O': [], 'Mn-O': []}
    with open(file, 'r') as f:
        for line in f:
            parts = line.split()
            if parts[0] in ['Ni-O', 'Co-O', 'Mn-O']:
                pair = parts[0]
                bonds = list(map(float, parts[1:]))
                bond_data[compound][pair].extend(bonds)
    for pair in ['Ni-O', 'Co-O', 'Mn-O']:
        averages[pair].append(np.mean(bond_data[compound][pair]))

# Scatter plot for bond lengths
colors  = {'Ni-O': '#ff7f0e', 'Co-O': '#d62728', 'Mn-O': '#1f77b4'}
offsets = {'Ni-O': 0.0, 'Co-O': 0.0, 'Mn-O': 0.0}
plt.figure(figsize=(11.2, 5.6))
plt.gca().yaxis.set_major_formatter(ticker.FormatStrFormatter('%3.1f'))
plt.gca().yaxis.set_major_locator(ticker.MultipleLocator(0.1))
plt.gca().yaxis.set_minor_locator(ticker.MultipleLocator(0.05))
for pair in ['Ni-O', 'Co-O', 'Mn-O']:
    data_for_pair = []
    positions = []
    for i, compound in enumerate(compound_labels):
        data_for_pair.append(bond_data[compound][pair])
        positions.append(i + 1 + offsets[pair])
    vp = plt.violinplot(data_for_pair, positions=positions, widths=0.4, showmeans=False, showmedians=False, showextrema=False)
    for body in vp['bodies']:
        body.set_facecolor(colors[pair])
        body.set_edgecolor('black')
        body.set_alpha(0.75)
legend_elements = [Patch(facecolor=colors['Ni-O'], edgecolor='black', label='Ni-O'),
                   Patch(facecolor=colors['Co-O'], edgecolor='black', label='Co-O'),
                   Patch(facecolor=colors['Mn-O'], edgecolor='black', label='Mn-O'),]

# Plot and connect average values for each compound (0K, 50K, 600K, SQS)
compound_indices = {'NCM523': [0, 1, 2, 3], 'NCM622': [4, 5, 6, 7], 'NCM811': [8, 9, 10, 11]}
for pair in ['Ni-O', 'Co-O', 'Mn-O']:
    x_vals = [i + 1 + offsets[pair] for i in range(len(compound_labels))]
    plt.scatter(x_vals, averages[pair], color='black', s=50, zorder=5)
for pair in ['Ni-O', 'Co-O', 'Mn-O']:
    for compound, indices in compound_indices.items():
        x_vals = [i + 1 + offsets[pair] for i in indices]
        y_vals = [averages[pair][i] for i in indices]
        plt.plot(x_vals, y_vals, color='black', linewidth=1)

# Other plot decoration
plt.ylim([1.8 - 0.1/6, 2.3 + 0.1/6])
plt.xticks(range(1, len(compound_labels) + 1), formatted_labels, rotation=45, fontsize=18)
plt.yticks(fontsize=24)
plt.ylabel('M-O Bond Length (Å)', fontsize=26)
plt.legend(handles=legend_elements, loc='upper right', ncol=3, frameon=False, handlelength=0.8, fontsize=20)
plt.tight_layout()
plt.savefig('NCM_Bonds_0K_50K_600K_SQS_violin.png')
plt.show()
```
